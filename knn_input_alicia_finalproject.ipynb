```python
# Define the pipeline for the KNN
knn_pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("clf", KNeighborsClassifier(n_neighbors=88))
])

# Train the KNN model
knn_pipeline.fit(X_train, y_train)

def predict_cardio(user_input):
    user_df = pd.DataFrame(user_input, index=[0])
    prediction = knn_pipeline.predict(user_df)
    return 'Cardiovascular Disease' if prediction[0] == 1 else 'No Cardiovascular Disease'

def get_user_input():
    user_input = {
        #'age': int(input("Enter age (days since birth): ")),
        #'gender': int(input("Enter gender (1 for female, 2 for male): ")),
        #'height': float(input("Enter height (cm): ")),
        #'weight': float(input("Enter weight (kg): ")),
        'ap_hi': int(input("Enter systolic blood pressure (ap_hi): ")),
        #'ap_lo': int(input("Enter diastolic blood pressure (ap_lo): ")),
        'cholesterol': int(input("Enter cholesterol level (1: normal, 2: above normal, 3: well above normal): ")),
        'gluc': int(input("Enter glucose level (1: normal, 2: above normal, 3: well above normal): ")),
        'smoke': int(input("Do you smoke? (0: no, 1: yes): ")),
        'alco': int(input("Do you drink alcohol? (0: no, 1: yes): ")),
        'active': int(input("Are you physically active? (0: no, 1: yes): "))
    }
    return user_input

# Gather user input
user_input = get_user_input()

result = predict_cardio(user_input)
print("Prediction for the input values is:", result)
```
